```python
x1=int(input('Enter a number:'))
x2=int(input('Enter a number:'))
ops=input('Enter an operator:')

if ops=='add':
  print(x1+x2)
elif ops=='sub':
  print(x1-x2)
elif ops=='mult':
  print(x1*x2)
elif ops =='div':
  if x2==0:
    print('error')
  else:
    print(x1/x2)
elif ops =='rem':
    print(x1%x2)
else :
  print('Invalid operation')
```
